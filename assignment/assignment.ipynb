### Question 1

Question: Write a Python function that takes a PySpark DataFrame `df` and a column name `col` as input, and returns a new DataFrame with a new column `new_col` that contains the square root of the values in `col`.

```python
import pandas as pd
from pyspark.sql import Row
```

```python
#Create a a PySpark DataFrame `df` and a column name `col`, with random data of type integer from 1 to 500 in multiples of 3
spark = SparkSession.builder.getOrCreate()
data = [(i,) for i in range(3, 501, 3)]
df = spark.createDataFrame(data, ["col"])
col = "col"
df.show(5)
# Write a python function that the df 'col' and returns a new DataFrame with a new column `new_col` 
# that contains the square root of the values in `col`.
def add_sqrt_column(df, col):
    from pyspark.sql.functions import sqrt
    new_df = df.withColumn("new_col", sqrt(col))
    return new_df
new_df = add_sqrt_column(df, col)
new_df.show(5)
```

### Question 3

Question: Write a PySpark code snippet to filter a DataFrame `df` to only include rows where the value in the `column1` column is greater than 10 and the value in the `column2` column contains the substring "foo".

```python
# create a dataframe with 2 columns 'column1' containing random intergers from 1 to 100 and 'column2' containing random strings of length 10 with some having substring 'foo' in it.
import random
import string
def random_string(length=10):
    letters = string.ascii_letters
    return ''.join(random.choice(letters) for _ in range(length))
data = []
for _ in range(100):
    num = random.randint(1, 100)
    if random.random() < 0.5:
        str_val = random_string()
    else:
        str_val = "foo" + random_string(7)
    data.append((num, str_val))
df_random = spark.createDataFrame(data, ["column1", "column2"])
df_random.show(10)
# filter the dataframe to only include rows where 'column1' is greater than 10 and 'column2' contains the substring 'foo'
from pyspark.sql.functions import col
filtered_df = df_random.filter((col("column1") > 10) & (col("column2").contains("foo")))
filtered_df.show(10)
```
